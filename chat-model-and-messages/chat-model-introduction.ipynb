```python
from langchain_ollama import ChatOllama
llm = ChatOllama(
        model="llama3.2:3b", 
        temperature=0,         
                 )
```

```python
messages =[
    ("system", "You are a helpful assistant that translates English to French. Translate the following sentence into French."),
    ("human", "Translate: I love programming."),
]
ai_msg = llm.invoke(messages)
ai_msg
```
